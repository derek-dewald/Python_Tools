```python
import holidays

class TimeSeriesForecastPipeline:
    def __init__(
        self,
        date_col: str,
        target_col: str,
        country_holidays: str = "CA",
        province: str | None = None,
        lags: list[int] = None,
        rolling_windows: list[int] = None,
    ):
        self.date_col = date_col
        self.target_col = target_col
        self.country_holidays = country_holidays
        self.province = province
        self.lags = lags if lags is not None else [1, 7, 14, 21, 28]
        self.rolling_windows = rolling_windows if rolling_windows is not None else [7, 14, 28]

        s

        self.feature_cols = None
        self.train_df_ = None



    def _build_features(self, df: pd.DataFrame) -> pd.DataFrame:
        df = self._prepare_base_df(df)
        df = self._add_calendar_features(df)
        df = self._add_lag_features(df)
        return df

    def fit(self, df: pd.DataFrame, train_end_date: str | None = None):
        df_feat = self._build_features(df)

        # drop rows where target is missing
        df_feat = df_feat[~df_feat[self.target_col].isna()].copy()

        # feature columns
        self.feature_cols = [
            c for c in df_feat.columns
            if c not in [self.date_col, self.target_col]
        ]

        # remove rows with NA features from lagging/rolling
        model_df = df_feat.dropna(subset=self.feature_cols).copy()

        if train_end_date is not None:
            train_end_date = pd.to_datetime(train_end_date)
            model_df = model_df[model_df[self.date_col] <= train_end_date].copy()

        X_train = model_df[self.feature_cols]
        y_train = model_df[self.target_col]

        self.model.fit(X_train, y_train)
        self.train_df_ = df.copy()

        return self

    def predict_historical(self, df: pd.DataFrame, test_start_date: str):
        df_feat = self._build_features(df)
        df_feat = df_feat[~df_feat[self.target_col].isna()].copy()

        test_start_date = pd.to_datetime(test_start_date)

        usable_df = df_feat.dropna(subset=self.feature_cols).copy()
        test_df = usable_df[usable_df[self.date_col] >= test_start_date].copy()

        X_test = test_df[self.feature_cols]
        test_df["prediction"] = self.model.predict(X_test)

        mae = mean_absolute_error(test_df[self.target_col], test_df["prediction"])
        rmse = np.sqrt(mean_squared_error(test_df[self.target_col], test_df["prediction"]))

        return test_df[[self.date_col, self.target_col, "prediction"]], {
            "MAE": mae,
            "RMSE": rmse
        }

    def predict_next_day(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Predict the next day after the latest available date.
        """
        df = self._prepare_base_df(df).copy()
        last_date = df[self.date_col].max()
        next_date = last_date + pd.Timedelta(days=1)

        # append empty next day row
        next_row = pd.DataFrame({
            self.date_col: [next_date],
            self.target_col: [np.nan]
        })
        temp_df = pd.concat([df, next_row], ignore_index=True)

        # build features
        feat_df = self._build_features(temp_df)

        pred_row = feat_df[feat_df[self.date_col] == next_date].copy()

        X_pred = pred_row[self.feature_cols]
        pred_value = self.model.predict(X_pred)[0]

        return pd.DataFrame({
            self.date_col: [next_date],
            "prediction": [pred_value]
        })


# Day of Week
# Day of Month
# Month
# Season
# Weekday
# Weekend
# holiday

# Box Plot by Each of the Variables.

~np.isclose(
        d['BRANCH_7D_CHANGE'],
        d['MBR_7D_CHANGE'],
        equal_nan=True
    )

def binary_complex_equivlance(df, col, col1, new_column_name):
    try:
        # Try numeric comparison
        df[new_column_name] = np.where(
            (df[col].isna() & df[col1].isna()) |
            ((df[col].fillna(0) == 0) & df[col1].isna()) |
            ((df[col1].fillna(0) == 0) & df[col].isna()) |
            (df[col].fillna(0) == df[col1].fillna(0)),
            1, 0
        )
    except Exception:
        # Fallback to string comparison
        df[new_column_name] = np.where(
            (df[col].isna() & df[col1].isna()) |
            ((df[col].fillna('') == '') & df[col1].isna()) |
            ((df[col1].fillna('') == '') & df[col].isna()) |
            (df[col].fillna('').astype(str).str.strip().str.lower() ==
             df[col1].fillna('').astype(str).str.strip().str.lower()),
            1, 0
        )
    return df


def df_column_comparison_binary(
    df,
    df1,
    primary_key_list,
    column_list=[]):

    d = df.merge(df1,on=primary_key_list,how='outer',suffixes=("","_"))

    if len(column_list)==0:
        column_list = d.columns
    
    for col in column_list:
        if (col in primary_key_list)| (col[-1]=='_') :
            pass
        else:
            try:
                df = binary_complex_equivlance(d,col,f"{col}_",f"{col}_FLAG")
            except:
                print(f'Column:{col} is Missing from One of the DF')


def column_comparison_df(
    df,
    primary_key,
    indicator='_'
):
    for col in df.columns:      
        if (col in primary_key)|(col[-1:]=='_')|(col=='_merge')|(col.find('_FLAG')!=-1):
            #print(col,'Passed')
            pass
        else:
            #print(col,'Not Passed')
            column_comparison(df,col,primary_key=primary_key)
    df['TOTAL_FLAGS']= r[[x for x in r.columns if x.find('_FLAG')!=-1]].sum(axis=1)
    


def search_missing_date(date_list,freq='d'):
    all_dates = pd.date_range(
        start=date_list.min(),
        end=date_list.max(),
        freq=freq)

    return all_dates.difference(date_list)

df['MONTH_WON'] = pd.NaT
mask = df['statuscode'] == 3
df.loc[mask, 'MONTH_WON'] = (
    df.loc[mask,'actualclosedate']
      .dt.to_period('M')
      .dt.to_timestamp('M')
)

with pd.ExcelWriter(filename, engine="xlsxwriter") as writer:
    # ---- Write the dataframes ----
    df.to_excel(writer, index=False, sheet_name="ALL_ACCOUNTS_OPENED")
    df1.to_excel(writer, index=False, sheet_name="SAMPLE_NEW_RETAIL_MEMBERSHIPS")

    workbook = writer.book

    # ---- Define formats ----
    acct_format = workbook.add_format({'num_format': '0'})
    date_format = workbook.add_format({'num_format': 'yyyy-mm-dd'})

    wrap_center = workbook.add_format({
        'text_wrap': True,
        'align': 'center',
        'valign': 'vcenter'
    })

    # ============================================================
    # ✅ Format Sheet 1
    # ============================================================
    ws1 = writer.sheets["ALL_ACCOUNTS_OPENED"]

    # --- Auto-size + wrap/center all columns ---
    for idx, col in enumerate(df.columns):
        max_len = min(
            80,
            max(df[col].astype(str).map(len).max(), len(col)) + 2
        )
        ws1.set_column(idx, idx, max_len, wrap_center)

    # --- Override ACCT and DATE formats ---
    acct_col_1 = df.columns.get_loc("ACCTNBR")
    date_col_1 = df.columns.get_loc("OPEN_DATE")

    ws1.set_column(acct_col_1, acct_col_1, 20, acct_format)
    ws1.set_column(date_col_1, date_col_1, 15, date_format)

    # ============================================================
    # ✅ Format Sheet 2
    # ============================================================
    ws2 = writer.sheets["SAMPLE_NEW_RETAIL_MEMBERSHIPS"]

    # --- Auto-size + wrap/center all columns ---
    for idx, col in enumerate(sample_df.columns):
        max_len = min(
            80,
            max(sample_df[col].astype(str).map(len).max(), len(col)) + 2
        )
        ws2.set_column(idx, idx, max_len, wrap_center)

    # --- Override ACCT and DATE formats ---
    acct_col_2 = sample_df.columns.get_loc("ACCTNBR")
    date_col_2 = sample_df.columns.get_loc("OPEN_DATE")

    ws2.set_column(acct_col_2, acct_col_2, 20, acct_format)
    ws2.set_column(date_col_2, date_col_2, 15, date_format)

def generate_formated_excel_file(df,
                                 filename,
                                 sheet_name='Sheet1',
                                 wrap=True,
                                 align='center',
                                 valign='vcenter'):
    
    with pd.ExcelWriter(filename, engine="xlsxwriter") as writer:
        df.to_excel(writer, index=False, sheet_name=sheet_name)
        
        workbook  = writer.book
        worksheet = writer.sheets[sheet_name]
        wrap_center = workbook.add_format({
            'text_wrap': wrap,
            'align':align,
            'valign':valign 
        })
        
        
        for idx, col in enumerate(df.columns):
            max_len = min(80,max(df[col].astype(str).map(len).max(), len(col)) + 2)
            worksheet.set_column(idx, idx, max_len, wrap_center)

transform performs a groupby operation but returns a Series that is the same length as the original dataframe—aligning each group’s result back to each row.

df1['NEW'] = df1.groupby(['OLD1', 'OLD2'])['NUM'].transform('mean')



mask = df1["EXCLUDE"] == 0


df1["TRANSCRIPT_COUNT"] = (
    df[mask]
    .groupby("transcriptionid")["transcriptionid"]
    .transform("count")
    .reindex(df1.index, fill_value=0)
)

final_df['new_time_column'] = (
    pd.to_datetime(final_df['time_column'], errors='coerce').dt.date
    .astype("object")
    .where(final_df['time_column'].notna(), None)
)

d['COLUMN_NAME'].str.len().max()


won['DATE_COL'].dt.to_period('M').dt.to_timestamp('M')

final_df1['MONTH_END'] = pd.NaT
final_df1.loc[final_df1['CONDITION_COLUMN'] == 3, 'OG_COLUMN'] = (
    final_df1['DATE_COL']
      .dt.to_period('M')
      .dt.to_timestamp('M')
)


def read_txt_file(file_name,
                  path_name='',
                  encoding="utf-8",
                  errors="replace",
                  return_value = 'list'):
    
    '''
    
    
    '''

    filepath = f"{path_name}{file_name}"
    with open(filepath, "r", encoding=encoding, errors=errors) as f:
        lines = [line for line in f if line]
    
    if return_value=='str':
        text = ''
        text += ''.join(lines)
        return text 
            
    return lines

def compare_text_files(file,file1,
                       print_length=100,
                       validation_length=20,
                       confirmation_space=10000):
    
    '''
    
    
    
    
    '''
    
    
    print(f'Length File:  {len(file)}\nLength File1: {len(file1)}')
    
    
    # Compare. 
    # Does it matter how Long? If over 1M should I sample.
    
    
    test_length = max(len(file),len(file1))
    
    for count in range(test_length):
        if file[count]!=file1[count]:
            print(f'At Position {count} there is a noted Difference.\nFile 1 Text:{file[count:count+print_length]}\nFile 1 Text:{file1[count:count+print_length]}')
            return []
        elif count in range(confirmation_space,test_length,confirmation_space):
            print(f"\nValidating Output, At Position:{count}.\nFile: {file[count:count+validation_length]}\nFile1:{file1[count:count+validation_length]}\n")


def generate_dict_for_df(df,
                         dict_df,
                         column_name_list=[],
                         primary_key='COLUMN'):
    
    '''
    
    
    
    '''
    
    # Need to Determine Which columns will have a Dictionary Created. If none defined, it will occur for all.
    if column_name_list ==[]:
        column_name_list = df.columns.tolist()
        
    if len(dict_df)!=len(dict_df.drop_duplicates(primary_key)):
        raise ValueError('Dictionary has Duplication, Please review before proceeding with Merge')
    
    final_df = pd.DataFrame(column_name_list,columns=[primary_key])
    final_df = final_df.merge(dict_df,on=primary_key,how='left')
    
    return final_df


METRIC_TYPE = ['BALANCE' - SPOT
               ,'VALUE'  - During Period.
               'CHANGE' - Calculation of Multiple Values.
               ,'VOLUME',,'ATTRIBUTE','RATE','FLAG', 'SCORE']


def generate_read_me(df,dictionary_source=None,column='Word'):
    
    '''
    
    
    '''

    if not dictionary_source:
        from data_d_dicts import links
        dict_df = pd.read_csv(links['google_definition_csv'])
    else:
        dict_df = pd.read_csv(dictionary_source)
    
    read_me_df = pd.DataFrame(df.columns,columns=[column])

    return read_me_df.merge(dict_df,on=column,how='left')


test = generate_read_me(df2)

def create_data_dictionary(df,
                           column_name_list=[],
                           dict_=pd.DataFrame(),
                           table_name=None,
                           dict_columns = ['Process','Categorization','Word','Definition'],
                           primary_key='Word',
                           include_frequency_count=False):
    
    '''
    
    '''
    # Need to Determine Which columns will have a Dictionary Created. If none defined, it will occur for all.
    if column_name_list ==[]:
        column_name_list = df.columns.tolist()

    # Need to determine source of Dictionary. As there could be multiple sources, need to be flexible.
    if len(dict_)==0:
        dict_ = pd.read_csv(links['google_definition_csv'])
        dict_ = dict_[dict_columns]

    # Create Baseline File, which will have items iteratively added to it

    final_df = pd.DataFrame()
    
    for word in column_name_list:
        temp_df = pd.DataFrame(df[word].unique(),columns=[primary_key])
        temp_df = temp_df.merge(dict_,on=primary_key,how='left')
        temp_df['Table Name'] = table_name
        temp_df['Column Name'] = word
        
        if include_frequency_count:
            temp_df['Frequency'] = temp_df['Word'].map(df.groupby(word)[word].size())        

        final_df = pd.concat([final_df,temp_df])

    return final_df

create_data_dictionary(df,
                       column_name_list=['Process','Categorization'],
                       table_name='D Learning Notes',
                       include_frequency_count=True,
                       dict_=dict_)


df = pd.read_csv(links['d_learning_notes_url'])
df.drop(['PRO_ORDER','CAT_ORDER'],axis=1,inplace=True)
df


def CreatePivotTableFromTimeSeries(df,
                                   index,
                                   date_column,
                                   values,
                                   metric={'7D':7,'1M':30,'3M':90,'6M':180,'12M':365},
                                   aggfunc='sum',
                                   t_value_dict=None,
                                   skipna=True):
    
    '''
    Function to Summaryize a Time Series Dataframe into a Pivot. Creating a number of critical Metrics.

    '''
    
    if not t_value_dict:
        t_value_dict = {
            'observed_col':'MEAN_1M',
            'mean_col':'MEAN_ALL',
            'std_col':'STD_DEV_ALL',
            'observations':'OBS'
        }           
    
    
    # 1. Pivot
    if index==None:
        df1 = df.pivot_table(columns=date_column,values=values,aggfunc=aggfunc).fillna(0)
    else:
        df1 = df.pivot_table(index=index, columns=date_column, values=values, aggfunc=aggfunc).fillna(0)
        
    # 2. Capture original month columns IMMEDIATELY after pivot
    month_cols = df[date_column].sort_values().unique().tolist()
    
    unique_dates = len(month_cols)
    
    df2 = df1[[df1.columns[-1]]].copy()
       
    for metric, period in metric.items():     
        if period<unique_dates:
            df2[f'MEAN_{metric}'] = df1[month_cols[-period:]].mean(axis=1)
            df2[f'STD_DEV_{metric}'] = df1[month_cols[-period:]].std(axis=1)
            
        else:
            df2[f'MEAN_ALL'] = df1[month_cols].mean(axis=1)
            df2[f'STD_DEV_ALL'] = df1[month_cols].std(axis=1)
            completed = True
            
    if not completed:
        df2['MEAN_ALL'] = df1[month_cols].mean(axis=1)
        df2['STD_DEV_ALL'] = df1[month_cols].mean(axis=1)
        
    try:
        df2['OBS'] = len(month_cols)
        df2 = t_value_test(df2,**t_value_dict)    
    
    except:
        pass
    
    df2['MAX_ALL'] = df1[month_cols].max(axis=1,skipna=True)
    df2['MIN_ALL'] = df1[month_cols].min(axis=1,skipna=True)
    df2['MEAN_CHG'] = df2[t_value_dict['observed_col']].fillna(0)-df2[t_value_dict['mean_col']].fillna(0)
    df2['MEAN_CHG_PERC'] = abs(df2['MEAN_CHG'].fillna(0)/df2[t_value_dict['mean_col']].fillna(0))
    
    total = df2[df1.columns[-1]].sum()
                
    df2['PERC_TOTAL'] = df2[df1.columns[-1]]/total
    df2['FLAG_REVIEW'] = np.where((df2['reject_null']==1)&(df2['MEAN_CHG_PERC']>.15)&(df2['PERC_TOTAL']>.005),1,0)
    df2['FLAG_REVIEW'] = np.where((df2['reject_null']==1)&(df2['MEAN_CHG_PERC']>.25)&(df2['PERC_TOTAL']>.001),1,df2['FLAG_REVIEW'])
    
    return df1.reset_index(),df2.reset_index()

import numpy as np
import pandas as pd
from scipy import stats

def t_value_test(
    df,
    observed_col,
    mean_col,
    std_col,
    observations,
    confidence_level= 0.90,
    alternative= "two-sided",   # 'two-sided', 'greater', or 'less'
    label= None,
    add_ci= False):
    
    """
    Row-wise one-sample t test of an observed value against a reference mean, with CI.

    Parameters
    ----------
    df : DataFrame
        Input data.
    observed_col : str
        Column with the observed value being tested.
    mean_col : str
        Column with the historical/sample mean (x̄).
    std_col : str
        Column with the sample standard deviation (s). Should be ddof=1.
    observations : str
        Column with the sample size (n).
    confidence_level : float, default=0.90
        CI level and test level (1 - alpha).
    alternative : {'two-sided','greater','less'}, default='two-sided'
        Type of test: two-sided or one-sided.
    label : str or None
        Suffix label to append to output column names (e.g., '90').
    add_ci : bool, default=True
        If True, add CI for the mean and inside/outside indicators.

    Returns
    -------
    DataFrame
        Copy of df with added columns: t_stat, t_crit, reject_null, and optionally ci_low/ci_high.
    """
    out = df.copy()

    alpha = 1.0 - confidence_level
    n = out[observations].astype(float)
    s = out[std_col].astype(float)
    se = s / np.sqrt(n)
    dfree = n - 1

    # Observed t-statistic: (obs - mean) / SE
    t_stat = (out[observed_col].astype(float) - out[mean_col].astype(float)) / se

    # Critical t-value(s), vectorized by df
    if alternative == "two-sided":
        t_crit = stats.t.ppf(1 - alpha/2.0, dfree)
        reject = t_stat.abs() > t_crit
    elif alternative == "greater":
        t_crit = stats.t.ppf(1 - alpha, dfree)
        reject = t_stat > t_crit
    elif alternative == "less":
        t_crit = stats.t.ppf(alpha, dfree)
        reject = t_stat < t_crit
    else:
        raise ValueError("alternative must be 'two-sided', 'greater', or 'less'.")

    suf = "" if label is None else f"_{label}"

    out[f"t_stat{suf}"] = t_stat
    
    out[f"reject_null{suf}"] = reject.astype("Int64")  # keeps NA if any inputs are NA

    if add_ci:
        out[f"t_crit{suf}"] = t_crit
        # CI for the population mean: x̄ ± tcrit * SE
        ci_margin = t_crit * se
        out[f"ci_low{suf}"] = out[mean_col] - ci_margin
        out[f"ci_high{suf}"] = out[mean_col] + ci_margin

        inside = (out[observed_col] >= out[f"ci_low{suf}"]) & (out[observed_col] <= out[f"ci_high{suf}"])
        out[f"outside_ci{suf}"] = (~inside).astype("Int64")

    return out
```

```python


import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    """
    Compute the sigmoid of x

    Arguments:
    x -- A scalar or numpy array of any size

    Return:
    s -- sigmoid(x)
    """
    
    s = 1/(1+np.exp(-x))
    
    return s


# Generate values for x
x = np.linspace(-10, 10, 200)

# Compute sigmoid values
y = sigmoid(x)

# Plot the curve
plt.figure()
plt.plot(x, y)

# Labels and title
plt.title("Sigmoid Function")
plt.xlabel("x")
plt.ylabel("sigmoid(x)")

# Optional grid
plt.grid(True)

plt.show()
```
