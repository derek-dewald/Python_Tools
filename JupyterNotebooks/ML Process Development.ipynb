```python
df = pd.read_excel('dict_blank.xlsx')

def apply_dictionary(
    df,
    new_column_name,
    mapping_dict,
    default_value='TBD',
    column_name="COLUMN_NAME",
):
    df = df.copy()
    conditions = [
        df[column_name].str.contains(key, na=False)
        for key in mapping_dict
    ]

    df[new_column_name] = np.select(
        conditions,
        mapping_dict.values(),
        default=default_value
    )
    return df

product_map = {
    'DATE': 'Not a Product',
    'CK': 'Chequing',
    'SAV': 'Savings',
    'TD': 'Term Deposit',
    'LOAN': 'Loan',
    'LOC': 'LOC',
    'TFSA': 'TFSA',
    'RRSP': 'RRSP',
    'HISA': 'HISA',
    'FHSA': 'FHSA',
    'REGISTERED': 'Registered',
    'BP': 'Bill Payment',
    'BC': 'Bank Card',
    'DEBIT': 'Debit',
    'CREDIT': 'Credit',
    'EMAIL': 'Email',
    'INTERNET': 'Internet',
    'EFT': 'EFT',
    'TXN': 'Transaction',
}

df = apply_dictionary(df,'PRODUCT',product_map,'Not a Product')

metric_dict = {
    'VAL_TOTAL':'Value',
    'VAL_AVG':'Average Value',
    'VAL_DIFF':'Difference Value',
    'VOL_TOTAL':'Count',
    'VOL_AVG':'Average Count',
    'VOL_DIFF':'Difference Count',
    'BAL_TOTAL':"Balance",
    'BAL_AVG':'Average Balance',
    'BAL_DIFF':'Difference Balance',
    'BAL':"Balance",
    'VAL':'Value',
    'VOL':"Volumne"
}

df = apply_dictionary(df,'METRIC',metric_dict,'Not a Metric')
df['METRIC'] = np.where((df['PRODUCT']!='Not a Product')&(df['METRIC']=='Not a Metric'),'Count',df['METRIC'])


time_dict = {
    '1M':'One Month Ago',
    '3M':'Three Months Ago',
    '6M':'Six Months Ago',
    '12M':'One Year Ago',
}

df = apply_dictionary(df,'TIME_FRAME',time_dict,'Not a Time Period')
df['TIME_FRAME'] = np.where((df['PRODUCT']!='Not a Product')&(df['TIME_FRAME']=='Not a Time Period'),'Spot',df['TIME_FRAME'])

df.to_excel('dict_blank_1.xlsx',index=False)

df[df['TIME_FRAME']=='Not a Time Period']
```

```python
def classify_breadth(row):
    product = row['PRODUCT']
    metric = row['METRIC']
    semantic = row['SEMANTIC_TYPE']
    column = row['COLUMN_NAME']

    # Product/account presence or count tells us about relationship breadth
    if semantic == 'Structural' and product not in ['Not a Product', None]:
        return 1

    # Explicit product/service/channel indicators
    if metric == 'Count' and product not in ['Not a Product', None]:
        return 1

    return 0

def classify_depth(row):
    metric = row['METRIC']
    semantic = row['SEMANTIC_TYPE']
    product = row['PRODUCT']

    depth_metrics = [
        'Balance',
        'Average Balance',
        'Value',
        'Average Value'
    ]

    if metric in depth_metrics and semantic in [
        'State',
        'Descriptive'
    ]:
        return 1

    return 0

def classify_behaviour(row):
    metric = row['METRIC']
    product = row['PRODUCT']

    behaviour_metrics = [
        'Volumne',
        'Count',
        'Average Count',
        'Value',
        'Average Value'
    ]

    behaviour_products = [
        'Debit',
        'Credit',
        'Transaction',
        'EFT',
        'Internet',
        'Email',
        'Bill Payment',
        'Bank Card'
    ]

    if product in behaviour_products and metric in behaviour_metrics:
        return 1

    return 0

def classify_trajectory(row):
    semantic = row['SEMANTIC_TYPE']

    trajectory_types = [
        'Diagnostic: Change',
        'Diagnostic: Trend',
        'Diagnostic: Momentum',
        'Diagnostic: Volatility'
    ]

    return int(semantic in trajectory_types)

def relationship_definitions(df):
    
    df['BREADTH'] = df.apply(classify_breadth, axis=1)
    df['DEPTH'] = df.apply(classify_depth, axis=1)
    df['BEHAVIOUR'] = df.apply(classify_behaviour, axis=1)
    df['TRAJECTORY'] = df.apply(classify_trajectory, axis=1)

    columns = ['BREADTH', 'DEPTH', 'BEHAVIOUR', 'TRAJECTORY']
    
    df['ENGAGEMENT_DEFINED'] = df[columns].sum(axis=1).gt(0).astype(int)
    df['ENGAGEMENT_DEF_COUNT'] = df[columns].sum(axis=1)

    gb = df[columns].groupby(columns).size().reset_index().rename(columns={0:"COUNT"})
    print(f'Distribution of Variables in Model to Engagement Definition:\n{gb}')
```

```python
import textwrap

def print_dict(d, indent=0, width=100):
    for key, value in d.items():
        prefix = "    " * indent

        if isinstance(value, dict):
            print(f"\n{prefix}{key}")
            print_dict(value, indent + 1, width)

        elif isinstance(value, list):
            print(f"\n{prefix}{key}")
            for item in value:
                text = str(item).strip()
                print(textwrap.fill(
                    text,
                    width=width,
                    initial_indent=f"{prefix}    - ",
                    subsequent_indent=f"{prefix}      "
                ))

        else:
            print(f"\n{prefix}{key}")
            text = str(value).strip()
            print(textwrap.fill(
                text,
                width=width,
                initial_indent=f"{prefix}    ",
                subsequent_indent=f"{prefix}    "
            ))

print_dict(project_dict)
```
